## Notebook Introduction: Understanding BDC Datasets for 2022 Q4

These files contain the core data for analyzing Business Development Company (BDC) investments for Q4 2022. Each file, formatted as TSV, plays a specific role in decoding the financials, disclosures, and taxonomy from SEC XBRL filings.

sub.tsv is the submissions file, with each row detailing a BDC filing. This includes metadata such as company name, location, filing date, CIK, fiscal period, and other key reporting attributes. Use this file to discover which filings are present and gather basic information about each submission.

tag.tsv defines every tag in the dataset, both standard and custom. It provides insight into what each tag represents, identifying whether it is custom, abstract, or related to a member or dimension. Refer to tag.tsv to understand the label or documentation for any tag found in num.tsv, txt.tsv, and other files.

num.tsv contains all the numerical data points from the BDC financial statements, such as asset values, investment values, principal amounts, and balances. This is the primary file for quantitative analysis.

txt.tsv includes textual disclosures and notes from filings, presenting plain text blocks like accounting policies, footnotes, and legal disclosures. This file is valuable for gaining context, narrative detail, or qualitative understanding from the filings.

pre.tsv explains how items were presented within each statement, indicating the line order, labels, and context such as grouping or statement type. Use pre.tsv to reconstruct the layout of a report or to match data items to their positions in statements.

cal.tsv contains information about calculation relationships between items, detailing how individual values are aggregated or related (for example, summing assets or outlining parent-child relationships).

non.tsv summarizes filings that are not financial statements, such as 8-K event filings. It provides essential details, tags, and values for these disclosures and is helpful for tracking supplementary or event-based metadata.

soi.tsv (Schedule of Investments) provides the most granular investment-level breakdown. Each row details a distinct portfolio holding reported by a BDC, including metadata (company, filing), sector, security type, investment specifics (interest rate, cost, fair value, maturity date, shares, gains/losses), and position within the portfolio. Use this file to analyze the composition and diversity of individual BDC fund holdings, including industry or issuer breakdowns and exposure to asset classes.

To use these files together, start by exploring sub.tsv to identify included filings. Conduct numerical analysis with num.tsv and use tag.tsv to interpret tags and their meanings. txt.tsv offers disclosure and policy text for deeper understanding, while pre.tsv helps you reconstruct report layouts. Use cal.tsv to trace or validate calculation and aggregation logic. Refer to non.tsv for supplementary or non-statement filings, and soi.tsv to study portfolio holdings and investment mix at the most detailed level.

```python
import pandas as pd

# File paths (now including soi.tsv)
base_path = r'C:\Users\pjayswal\Downloads\2022_Q4-bdc\2022_Q4\datasets\\'
file_names = [
    'num.tsv', 'txt.tsv', 'cal.tsv', 'non.tsv', 'pre.tsv', 'sub.tsv', 'tag.tsv', 'soi.tsv'
]

# Load all files into dataframes
dfs = {}
for file in file_names:
    path = base_path + file
    df = pd.read_csv(path, sep='\t', encoding='utf-8', low_memory=False)
    dfs[file.replace('.tsv', '')] = df

# Preview data shapes and missing values to start cleaning
for name, df in dfs.items():
    print(f'\n--- {name} ---')
    print('Shape:', df.shape)
    print('Missing values per column:')
    print(df.isnull().sum())
    print(df.head())
```

```python
# Data cleaning step 1: Fix common missing values, drop highly-null columns if needed, ensure types

for name, df in dfs.items():
    print(f'\n--- Cleaning: {name} ---')
    
    # Drop columns that are almost entirely missing (if >98% null), except key ID columns
    # Modify drop rules per your analysis needs
    null_col_thresh = 0.98
    to_drop = [col for col in df.columns 
               if df[col].isnull().mean() > null_col_thresh and col not in ['adsh', 'tag', 'version']]
    if to_drop:
        print(f'Dropping columns (mostly missing): {to_drop}')
        df.drop(columns=to_drop, inplace=True)
    
    # Fill common nulls:
    # Numeric columns: fill with 0.0 or leave as np.nan if computation needed
    for col in df.select_dtypes(include=['float', 'int']).columns:
        df[col].fillna(0, inplace=True)
    
    # Text columns: fill with empty string if needed
    for col in df.select_dtypes(include=['object']).columns:
        df[col].fillna('', inplace=True)
    
    # Show sample after cleaning
    print(df.info())
    print(df.head())
```

```python
from sklearn.linear_model import Ridge
from sklearn.model_selection import LeaveOneOut
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np
import pandas as pd

print("LEAVE-ONE-OUT CROSS-VALIDATION (LOOCV)")
print("="*80)
print("This tests TRUE out-of-sample performance (removes each sample once)")
print("="*80)

# Convert to DataFrame for proper indexing (FIX)
X_q4_key_df = X_q4_key.reset_index(drop=True)
y_change_valid_series = pd.Series(y_change_valid).reset_index(drop=True)
companies_list = list(X_q4_key.index)

# Setup LOOCV
loo = LeaveOneOut()

# Storage for predictions
y_pred_loo = np.zeros(len(y_change_valid))

# Run LOOCV
for train_index, test_index in loo.split(X_q4_key_df):
    X_train_loo = X_q4_key_df.iloc[train_index]
    y_train_loo = y_change_valid_series.iloc[train_index]
    
    X_test_loo = X_q4_key_df.iloc[test_index]
    
    # Fit model on 24 samples, test on 1
    reg_loo = Ridge(alpha=0.01)
    reg_loo.fit(X_train_loo, y_train_loo)
    
    # Predict the left-out sample
    y_pred_loo[test_index] = reg_loo.predict(X_test_loo)

# Evaluate
r2_loo = r2_score(y_change_valid, y_pred_loo)
mae_loo = mean_absolute_error(y_change_valid, y_pred_loo)
rmse_loo = np.sqrt(mean_squared_error(y_change_valid, y_pred_loo))

print(f"\nLOOCV Results (True Out-of-Sample Performance):")
print("="*80)
print(f"R² Score: {r2_loo:.4f}")
print(f"MAE: ${mae_loo:.4f}")
print(f"RMSE: ${rmse_loo:.4f}")

print(f"\n\nComparison: Training vs. LOOCV")
print("="*80)
comparison_df = pd.DataFrame({
    'Metric': ['R² Score', 'MAE', 'RMSE'],
    'Training (Fit on all 25)': [0.0511, 1.2879, 2.6887],
    'LOOCV (Leave-one-out)': [f"{r2_loo:.4f}", f"${mae_loo:.4f}", f"${rmse_loo:.4f}"],
})
print(comparison_df.to_string(index=False))

print(f"\n\nLOOCV Predictions vs. Actuals:")
print("="*80)
loocv_results = pd.DataFrame({
    'Company': companies_list,
    'Actual': y_change_valid,
    'LOOCV_Pred': y_pred_loo,
    'Error': y_change_valid - y_pred_loo,
    'Abs_Error': np.abs(y_change_valid - y_pred_loo)
}).sort_values('Abs_Error', ascending=False)

print(loocv_results.head(15).to_string(index=False))

print(f"\n\nSummary Statistics:")
print("="*80)
print(f"Mean Absolute Error: ${loocv_results['Abs_Error'].mean():.4f}")
print(f"Median Absolute Error: ${loocv_results['Abs_Error'].median():.4f}")
print(f"Best prediction error: ${loocv_results['Abs_Error'].min():.4f}")
print(f"Worst prediction error: ${loocv_results['Abs_Error'].max():.4f}")
```

```python
import pandas as pd
import numpy as np

print("="*80)
print("ENGINEERING HOLDINGS-LEVEL FEATURES FOR PREDICTIVE MODELING")
print("="*80)

# Fix date issue: Use 'period' column instead of 'ddate'
soi_combined['period'] = pd.to_datetime(soi_combined['period'], format='%Y%m%d', errors='coerce')

print(f"\nDate range (using 'period'): {soi_combined['period'].min()} to {soi_combined['period'].max()}")

# Select KEY value and metric columns for feature engineering
key_features = {
    'Investment Owned, Fair Value': 'sum',        # Total portfolio fair value
    'Investment Owned, Cost': 'sum',               # Total portfolio cost
    'Investment Owned, Net Assets, Percentage': 'mean',  # Avg % of net assets
    'Investment Interest Rate': 'mean',            # Avg interest rate (CRITICAL for BDC)
    'Investment, Basis Spread, Variable Rate': 'mean',  # Avg spread
    'Investment, Interest Rate, Floor': 'mean',   # Avg rate floor
    'Investment Owned, Balance, Principal Amount': 'sum',  # Total principal
    'Concentration Risk, Percentage': 'max',      # Max concentration (risk metric)
}

# Also count holdings and diversity
print("\nAggregating by BDC and quarter...")

# Group 1: Aggregate numeric features
agg_features = {}
for col, agg_func in key_features.items():
    if col in soi_combined.columns:
        agg_features[col] = agg_func

bdc_features = soi_combined.groupby(['name', 'quarter']).agg(agg_features).reset_index()

# Group 2: Count holdings and calculate diversity
holdings_count = soi_combined.groupby(['name', 'quarter']).size().reset_index(name='num_holdings')

# Group 3: Sector diversity
sector_diversity = soi_combined.groupby(['name', 'quarter'])['Industry Sector Axis'].nunique().reset_index(
    name='num_sectors'
)

# Group 4: Investment type diversity  
investment_type_diversity = soi_combined.groupby(['name', 'quarter'])['Investment Type Axis'].nunique().reset_index(
    name='num_investment_types'
)

# Merge all aggregations
bdc_features = bdc_features.merge(holdings_count, on=['name', 'quarter'])
bdc_features = bdc_features.merge(sector_diversity, on=['name', 'quarter'], how='left')
bdc_features = bdc_features.merge(investment_type_diversity, on=['name', 'quarter'], how='left')

# Fill NaNs
bdc_features = bdc_features.fillna(0)

print(f"\nEngineered features shape: {bdc_features.shape}")
print(f"Features created: {bdc_features.columns.tolist()}")

# Rename columns for clarity
bdc_features.columns = [
    'BDC_Name', 'Quarter',
    'portfolio_fair_value_sum',
    'portfolio_cost_sum',
    'avg_pct_net_assets',
    'avg_interest_rate',
    'avg_spread',
    'avg_rate_floor',
    'total_principal',
    'max_concentration_risk',
    'num_holdings',
    'num_sectors',
    'num_investment_types'
]

print("\n" + "="*80)
print("SUMMARY STATISTICS - Holdings-Level Features")
print("="*80)

print("\nQ4 2022 Sample (first 5 BDCs):")
print(bdc_features[bdc_features['Quarter'] == '2022Q4'].head())

print("\n\nQ1 2023 Sample (first 5 BDCs):")
print(bdc_features[bdc_features['Quarter'] == '2023Q1'].head())

print("\n\nFeature Statistics (Q4 2022):")
q4_stats = bdc_features[bdc_features['Quarter'] == '2022Q4'].describe()
print(q4_stats)

print("\n\nFeature Statistics (Q1 2023):")
q1_stats = bdc_features[bdc_features['Quarter'] == '2023Q1'].describe()
print(q1_stats)

print("\n\nMissing values:")
print(bdc_features.isnull().sum())
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import warnings
warnings.filterwarnings('ignore')

print("="*80)
print("LINEAR REGRESSION TIME SERIES WITH HOLDINGS-LEVEL REGRESSORS")
print("="*80)

# Prepare NAV time series data
nav_ts = nav_for_prophet_clean.copy()
nav_ts['date'] = nav_ts['quarter'].map({'2022Q4': '2022-10-31', '2023Q1': '2023-03-31'})
nav_ts['date'] = pd.to_datetime(nav_ts['date'])
nav_ts = nav_ts[['name', 'date', 'nav_per_share']].sort_values(['name', 'date'])

# Prepare holdings features
features_ts = bdc_features.copy()
features_ts['date'] = features_ts['Quarter'].map({'2022Q4': '2022-10-31', '2023Q1': '2023-03-31'})
features_ts['date'] = pd.to_datetime(features_ts['date'])
features_ts = features_ts.rename(columns={'BDC_Name': 'name'})

# Merge NAV + features
data_combined = nav_ts.merge(
    features_ts[['name', 'date', 'avg_interest_rate', 'avg_spread', 'max_concentration_risk', 
                  'num_holdings', 'num_sectors']],
    on=['name', 'date'],
    how='left'
)

print(f"Combined data (NAV + features): {data_combined.shape}")

# Store results
all_results = []

print(f"\nFitting Linear Regression models for {len(bdcs_both)} BDCs with regressors...\n")

for i, bdc in enumerate(sorted(bdcs_both), 1):
    try:
        # Get BDC time series
        ts_data = data_combined[data_combined['name'] == bdc].sort_values('date').copy()
        
        if len(ts_data) < 2:
            continue
        
        # Extract features
        y = ts_data[['nav_per_share']].values  # Target: NAV per share
        X = ts_data[['avg_interest_rate', 'avg_spread', 'num_holdings', 'num_sectors']].fillna(0).values
        
        # Fit linear regression
        model = LinearRegression()
        model.fit(X, y)
        
        # Predict
        y_pred = model.predict(X)
        
        # Get Q4 and Q1 actual values
        q4_actual = y[0, 0]
        q1_actual = y[1, 0]
        q1_pred = y_pred[1, 0]
        
        # Calculate metrics
        error = q1_actual - q1_pred
        mape = abs(error) / q1_actual * 100 if q1_actual != 0 else 0
        
        all_results.append({
            'BDC': bdc,
            'Q4_2022_Actual': q4_actual,
            'Q1_2023_Actual': q1_actual,
            'Q1_2023_Predicted': q1_pred,
            'Error': error,
            'Abs_Error': abs(error),
            'MAPE': mape,
            'NAV_Change': q1_actual - q4_actual
        })
        
        print(f"[{i:2d}/{len(bdcs_both)}] {bdc:45s} | Act: ${q1_actual:7.2f} → Pred: ${q1_pred:7.2f} | Error: ${error:+7.2f} ({mape:5.1f}%)")
        
    except Exception as e:
        print(f"[{i:2d}/{len(bdcs_both)}] {bdc:45s} | ERROR: {str(e)[:30]}")

# Results DataFrame
if len(all_results) > 0:
    results_df = pd.DataFrame(all_results).sort_values('Abs_Error')

    print(f"\n\n" + "="*80)
    print("LINEAR REGRESSION WITH REGRESSORS - RESULTS")
    print("="*80)

    print("\nTop 10 BEST Predictions (lowest error):")
    print(results_df.head(10).to_string(index=False))

    print("\n\nTop 10 WORST Predictions (highest error):")
    print(results_df.tail(10).to_string(index=False))

    print(f"\n\n" + "="*80)
    print("MODEL PERFORMANCE METRICS")
    print("="*80)

    mae = results_df['Abs_Error'].mean()
    median_error = results_df['Abs_Error'].median()
    mape = results_df['MAPE'].mean()
    rmse = np.sqrt((results_df['Error']**2).mean())
    r2 = r2_score(results_df['Q1_2023_Actual'], results_df['Q1_2023_Predicted'])

    print(f"Mean Absolute Error (MAE):        ${mae:.4f}")
    print(f"Median Absolute Error:            ${median_error:.4f}")
    print(f"Mean Absolute Percentage Error:   {mape:.2f}%")
    print(f"Root Mean Squared Error (RMSE):   ${rmse:.4f}")
    print(f"R² Score:                         {r2:.4f}")
    print(f"\nTotal BDCs predicted:             {len(results_df)}")
    print(f"Predictions within ±$1.00:        {(results_df['Abs_Error'] <= 1.0).sum()} ({(results_df['Abs_Error'] <= 1.0).sum()/len(results_df)*100:.1f}%)")
    print(f"Predictions within ±$2.00:        {(results_df['Abs_Error'] <= 2.0).sum()} ({(results_df['Abs_Error'] <= 2.0).sum()/len(results_df)*100:.1f}%)")
    print(f"Predictions within ±$0.50:        {(results_df['Abs_Error'] <= 0.5).sum()} ({(results_df['Abs_Error'] <= 0.5).sum()/len(results_df)*100:.1f}%)")

else:
    print("⚠️ No results - check data!")
```

```python
import pandas as pd
import numpy as np
from prophet import Prophet
import warnings
warnings.filterwarnings('ignore')

print("="*80)
print("PROPHET TIME SERIES: BDC REALIZED GAINS/LOSSES")
print("="*80)

# Define quarters
quarters = {
    '2022Q4': (r'C:\Users\pjayswal\Downloads\2022_Q4-bdc\2022_Q4\datasets\num.tsv',
               r'C:\Users\pjayswal\Downloads\2022_Q4-bdc\2022_Q4\datasets\sub.tsv'),
    '2023Q1': (r'C:\Users\pjayswal\Downloads\2023_Q1-20251029T072820Z-1-001\2023_Q1\datasets\num.tsv',
               r'C:\Users\pjayswal\Downloads\2023_Q1-20251029T072820Z-1-001\2023_Q1\datasets\sub.tsv'),
    '2023Q2': (r'C:\Users\pjayswal\Downloads\2023_Q2-20251029T205027Z-1-001\2023_Q2\datasets\num.tsv',
               r'C:\Users\pjayswal\Downloads\2023_Q2-20251029T205027Z-1-001\2023_Q2\datasets\sub.tsv'),
    '2023Q3': (r'C:\Users\pjayswal\Downloads\2023_Q3-20251029T205018Z-1-001\2023_Q3\datasets\num.tsv',
               r'C:\Users\pjayswal\Downloads\2023_Q3-20251029T205018Z-1-001\2023_Q3\datasets\sub.tsv'),
    '2023Q4': (r'C:\Users\pjayswal\Downloads\2023_Q4-20251029T204804Z-1-001\2023_Q4\datasets\num.tsv',
               r'C:\Users\pjayswal\Downloads\2023_Q4-20251029T204804Z-1-001\2023_Q4\datasets\sub.tsv')
}

dates_map = {
    '2022Q4': '2022-09-30',
    '2023Q1': '2022-12-31',
    '2023Q2': '2023-03-31',
    '2023Q3': '2023-06-30',
    '2023Q4': '2023-09-30'
}

# Load and extract gains
gains_data = {}
for qtr, (num_path, sub_path) in quarters.items():
    num_df = pd.read_csv(num_path, sep='\t', encoding='utf-8', low_memory=False)
    sub_df = pd.read_csv(sub_path, sep='\t', encoding='utf-8')
    
    gains = num_df[num_df['tag'] == 'NetInvestmentIncome'][['adsh', 'value']].copy()

    gains = gains.merge(sub_df[['adsh', 'name']], on='adsh')
    
    gains_data[qtr] = gains.groupby('name')['value'].mean().reset_index()
    gains_data[qtr]['date'] = dates_map[qtr]
    print(f"✓ {qtr}: {len(gains_data[qtr])} BDCs")

# Combine all
all_gains = pd.concat(gains_data.values(), ignore_index=True)
all_gains['date'] = pd.to_datetime(all_gains['date'])
all_gains = all_gains.sort_values(['name', 'date']).reset_index(drop=True)

print(f"\n✓ Total records: {len(all_gains)}")

# Build Prophet for each BDC
print("\nBuilding Prophet models...")

results = []
count = 0

for bdc in all_gains['name'].unique():
    try:
        bdc_data = all_gains[all_gains['name'] == bdc][['date', 'value']].copy()
        
        if len(bdc_data) < 2:
            continue
        
        # Remove NaN
        bdc_data = bdc_data.dropna()
        if len(bdc_data) < 2:
            continue
        
        # Prepare for Prophet
        bdc_data = bdc_data.rename(columns={'date': 'ds', 'value': 'y'})
        bdc_data['ds'] = pd.to_datetime(bdc_data['ds'])
        bdc_data = bdc_data.sort_values('ds').reset_index(drop=True)
        
        # Fit Prophet
        model = Prophet(yearly_seasonality=False, weekly_seasonality=False)
        with warnings.catch_warnings():
            warnings.simplefilter("ignore")
            model.fit(bdc_data)
        
        # Predict on same data
        forecast = model.predict(bdc_data[['ds']])
        
        # Compare
        for i in range(len(bdc_data)):
            actual = bdc_data.loc[i, 'y']
            predicted = forecast.loc[i, 'yhat']
            error = actual - predicted
            mape = abs(error) / abs(actual) * 100 if actual != 0 else 0
            
            results.append({
                'BDC': bdc,
                'Date': bdc_data.loc[i, 'ds'],
                'Actual': actual,
                'Predicted': predicted,
                'Error': error,
                'Abs_Error': abs(error),
                'MAPE': mape
            })
        
        count += 1
    
    except Exception as e:
        pass

print(f"✓ Prophet fitted: {count} BDCs")

# Results
if len(results) > 0:
    df_results = pd.DataFrame(results).sort_values('MAPE')
    
    print("\n" + "="*80)
    print("PROPHET RESULTS")
    print("="*80)
    
    print(f"\nTotal Forecasts: {len(df_results)}")
    print(f"Mean MAPE: {df_results['MAPE'].mean():.2f}%")
    print(f"Median MAPE: {df_results['MAPE'].median():.2f}%")
    print(f"Mean Abs Error: ${df_results['Abs_Error'].mean()/1e6:.2f}M")
    
    print("\nTop 10 Best:")
    print(df_results.head(10)[['BDC', 'Date', 'Actual', 'Predicted', 'MAPE']].to_string(index=False))
    
    print("\nTop 10 Worst:")
    print(df_results.tail(10)[['BDC', 'Date', 'Actual', 'Predicted', 'MAPE']].to_string(index=False))
    
    df_results.to_csv('bdc_prophet_results.csv', index=False)
    print("\n✓ Saved: bdc_prophet_results.csv")
else:
    print("No predictions generated")
```
